# Customer Churn Analysis

```python
import pandas as pd

df = pd.read_csv("/Users/aisyahihsani/Downloads/telco2.csv")
df.head()
```

```python
df.info()
```

## Data Cleaning

```python
(df['TotalCharges'] == " ").sum()
```

```python
df['TotalCharges'] = df['TotalCharges'].replace(" ", pd.NA)
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'])
```

```python
df.info()
```

```python
df['TotalCharges'].isna().sum()
```

```python
df = df.dropna()
#drop missing values 
```

```python
df.isna().sum()
```

## EXPLORATORY DATA ANALYSIS

```python
df['Churn'].value_counts(normalize=True) * 100
```

```python
pd.crosstab(df['Contract'], df['Churn'], normalize='index') * 100
```

```python
#is churning customer paying higher monthly charges?

df.groupby('Churn')['MonthlyCharges'].mean()
```

Customers who churn tend to have higher average monthly charges compared to those who stay, suggesting that pricing may be a key factor influencing churn. High-paying customers may feel that the service is not worth the cost, leading to higher churn rates.

```python
#is new customer will likely to churn?

df.groupby('Churn')['tenure'].mean()
```

```python
pd.crosstab(df['InternetService'], df['Churn'], normalize='index') * 100
```
